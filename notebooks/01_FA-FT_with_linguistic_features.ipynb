```python
"""
=============================================================
  AES — Paradigma A: IndoSBERT-large Fine-tuning + FE Augmentation
  Dataset  : preprocessed_texts.pkl, scores.pkl
  Model    : denaya/indoSBERT-large
  Arsitektur:
    (0) Transformer: BertModel LARGE — 24 layer, hidden 1024
    (1) Pooling    : mean pooling (pooling_mode_mean_tokens=True)
    (2) Dense      : Linear(1024→256) + Tanh  → output dim 256
  FC head   : Linear(256+n_feat → 128) + ReLU + Dropout + Linear(128→1)
  Metrik   : QWK, MSE
  Evaluasi : 10-Fold Cross-Validation
=============================================================
"""

import os, time, traceback, random
from datetime import datetime

import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import pynvml
from torch.utils.data import Dataset, DataLoader
from sentence_transformers import SentenceTransformer
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import KFold
# ── [FIX-LEAKAGE] inner early-stopping split ────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

ES_FRAC        = 0.10   # share of the outer-train fold used for early stopping
SCALE_FEATURES = True   # FA-FT only: z-score linguistic features (fit on train only)

def inner_split(train_idx, y, fold_idx, frac=ES_FRAC, seed=42):
    """Split the outer-train fold into (train, early-stop) sets.
    Stratified by score band (scores <=5 merged) so rare low scores appear in both.
    Same seed in every notebook -> identical splits for FA-FT and FA-ML."""
    bands = np.clip(np.asarray(y)[train_idx].astype(int), 5, 10)
    tr, es = train_test_split(train_idx, test_size=frac,
                              random_state=seed + fold_idx, stratify=bands)
    return np.sort(tr), np.sort(es)
# ────────────────────────────────────────────────────────────────────────
from sklearn.metrics import mean_squared_error
from nltk.tokenize import word_tokenize, sent_tokenize
from nlp_id.postag import PosTag


# ══════════════════════════════════════════════════════════
#  KONFIGURASI
# ══════════════════════════════════════════════════════════
CFG = {
    "essay"        : f"{ESSAY_PATH}preprocessed_sbert_texts.pkl",
    "score"        : f"{SCORE_PATH}scores.pkl",
    "question_ids" : f"{SCORE_PATH}question_ids.pkl",
    "model_name"      : "denaya/indoSBERT-large",
    "max_len"         : 256,
    "sbert_output_dim": 256,
    "lr_encoder"      : 2e-5,
    "lr_head"         : 1e-3,
    "weight_decay"    : 0.01,
    "warmup_steps"    : 100,
    "max_epochs"      : 30,
    "batch_size"      : 64,
    "patience"        : 5,
    "n_folds"         : 10,
    "seed"            : 42,
    "output_dir"      : f"{OUTPUT_PATH}output_indosbert_ft",
    "score_min"       : 2,
    "score_max"       : 10,
}

os.makedirs(CFG["output_dir"], exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device           : {DEVICE}")
print(f"Model            : {CFG['model_name']}")
print(f"Backbone         : BERT-large (24L, hidden 1024)")
print(f"Output embedding : {CFG['sbert_output_dim']} dim (setelah Dense Tanh)")
print(f"do_lower_case    : False")


# ══════════════════════════════════════════════════════════
#  REPRODUCIBILITY
# ══════════════════════════════════════════════════════════
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ══════════════════════════════════════════════════════════
#  LINGUISTIC FEATURE EXTRACTION
#  6 fitur — dihitung dari teks preprocessed yang mempertahankan
#  tanda baca (bukan dari raw text terpisah).
#
#  Fitur:
#  1. ttr              — lexical diversity (unique/total tokens)
#  2. word_count       — panjang respons, proxy cakupan konten
#  3. noun_count       — kepadatan konten faktual (NN, NNP)
#  4. verb_count       — elaborasi naratif/kausal (VB)
#  5. sentence_count   — kompleksitas sintaksis (proxy)
#  6. avg_word_length  — kompleksitas morfologi (proxy, Indonesian-specific)
# ══════════════════════════════════════════════════════════
postag = PosTag()

POS_NOUN = {'NN', 'NNP'}
POS_VERB = {'VB'}

FEATURE_NAMES = [
    'ttr',
    'word_count',
    'noun_count',
    'verb_count',
    'sentence_count',
    'avg_word_length',
]


def extract_linguistic_features(text: str) -> dict:
    if not text:
        return {f: 0.0 for f in FEATURE_NAMES}

    words       = word_tokenize(text)
    word_tokens = [w for w in words if w.isalnum()]
    n_words     = len(word_tokens)

    if n_words == 0:
        return {f: 0.0 for f in FEATURE_NAMES}

    ttr             = len(set(word_tokens)) / n_words
    word_count      = float(n_words)
    pos_tags        = postag.get_pos_tag(text)
    noun_count      = sum(1 for _, pos in pos_tags if pos in POS_NOUN)
    verb_count      = sum(1 for _, pos in pos_tags if pos in POS_VERB)
    sentences       = sent_tokenize(text)
    sentence_count  = float(max(len(sentences), 1))
    avg_word_length = sum(len(w) for w in word_tokens) / n_words

    return {
        'ttr'            : ttr,
        'word_count'     : word_count,
        'noun_count'     : float(noun_count),
        'verb_count'     : float(verb_count),
        'sentence_count' : sentence_count,
        'avg_word_length': avg_word_length,
    }


# ══════════════════════════════════════════════════════════
#  QWK
# ══════════════════════════════════════════════════════════
def quadratic_weighted_kappa(y_true, y_pred, min_rating, max_rating):
    y_true = np.array(y_true, dtype=int)
    y_pred = np.round(np.clip(y_pred, min_rating, max_rating)).astype(int)
    num_ratings = max_rating - min_rating + 1
    conf_mat    = np.zeros((num_ratings, num_ratings), dtype=int)
    for a, p in zip(y_true, y_pred):
        conf_mat[a - min_rating][p - min_rating] += 1
    num_scored = len(y_true)
    hist_true  = np.bincount(y_true - min_rating, minlength=num_ratings)
    hist_pred  = np.bincount(y_pred - min_rating, minlength=num_ratings)
    expected   = np.outer(hist_true, hist_pred) / num_scored
    weights    = np.array([[(i - j)**2 / (num_ratings - 1)**2
                             for j in range(num_ratings)]
                            for i in range(num_ratings)])
    denom = np.sum(weights * expected)
    return 0.0 if denom == 0 else 1.0 - np.sum(weights * conf_mat) / denom


# ══════════════════════════════════════════════════════════
#  DATASET
# ══════════════════════════════════════════════════════════
class EssayDataset(Dataset):
    def __init__(self, texts, features, scores):
        self.texts    = texts
        self.features = features
        self.scores   = scores

    def __len__(self): return len(self.texts)

    def __getitem__(self, idx):
        return {
            "text"    : self.texts[idx],
            "features": torch.tensor(self.features[idx], dtype=torch.float32),
            "score"   : torch.tensor(self.scores[idx],   dtype=torch.float32),
        }


def collate_fn(batch):
    return {
        "texts"   : [b["text"]     for b in batch],
        "features": torch.stack([b["features"] for b in batch]),
        "scores"  : torch.stack([b["score"]    for b in batch]),
    }


# ══════════════════════════════════════════════════════════
#  MODEL — IndoSBERTRegressor
# ══════════════════════════════════════════════════════════
class IndoSBERTRegressor(nn.Module):
    def __init__(self, model_name, n_feat, device):
        super().__init__()
        self.sbert  = SentenceTransformer(model_name, device=str(device))
        emb_dim     = CFG["sbert_output_dim"]
        self.n_feat = n_feat
        self.dropout = nn.Dropout(0.1)
        self.fc1    = nn.Linear(emb_dim + n_feat, 128)
        self.relu   = nn.ReLU()
        self.fc2    = nn.Linear(128, 1)

    def encode_with_grad(self, texts):
        tokenizer = self.sbert.tokenizer
        features  = tokenizer(
            texts, padding=True, truncation=True,
            max_length=CFG["max_len"], return_tensors="pt")
        dev      = next(self.sbert.parameters()).device
        features = {k: v.to(dev) for k, v in features.items()}

        transformer_out = self.sbert[0].auto_model(**features)
        token_emb       = transformer_out.last_hidden_state   # (B, seq_len, 1024)

        mask     = features["attention_mask"].unsqueeze(-1).float()
        sum_emb  = torch.sum(token_emb * mask, dim=1)
        sum_mask = torch.clamp(mask.sum(dim=1), min=1e-9)
        mean_emb = sum_emb / sum_mask                         # (B, 1024)

        out = self.sbert[2]({"sentence_embedding": mean_emb})
        return out["sentence_embedding"]                      # (B, 256)

    def forward(self, texts, features):
        emb    = self.encode_with_grad(texts)
        emb    = self.dropout(emb)
        concat = torch.cat([emb, features], dim=1)
        return self.fc2(self.dropout(self.relu(self.fc1(concat)))).squeeze(-1)


# ══════════════════════════════════════════════════════════
#  TRAINING & EVALUATION
# ══════════════════════════════════════════════════════════
def train_one_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    for batch in loader:
        optimizer.zero_grad()
        preds = model(batch["texts"], batch["features"].to(device))
        loss  = criterion(preds, batch["scores"].to(device))
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
    return total_loss / len(loader)


def evaluate(model, loader, device, score_min, score_max):
    model.eval()
    all_preds, all_true, total_loss = [], [], 0.0
    criterion = nn.MSELoss()
    with torch.no_grad():
        for batch in loader:
            preds = model(batch["texts"], batch["features"].to(device))
            total_loss += criterion(preds, batch["scores"].to(device)).item()
            all_preds.extend(preds.cpu().numpy().tolist())
            all_true.extend(batch["scores"].cpu().numpy().tolist())
    all_preds = np.array(all_preds)
    all_true  = np.array(all_true)
    return (total_loss / len(loader),
            mean_squared_error(all_true, all_preds),
            quadratic_weighted_kappa(all_true, all_preds, score_min, score_max),
            all_preds, all_true)


# ══════════════════════════════════════════════════════════
#  RUN FOLD
# ══════════════════════════════════════════════════════════
def run_fold(fold_idx, train_idx, val_idx, texts, features_raw,
             scores, question_ids, cfg, device, es_idx=None):

    print(f"\n{'='*55}\n  FOLD {fold_idx+1}/{cfg['n_folds']}\n{'='*55}")
    set_seed(cfg["seed"] + fold_idx)
    if device.type == "cuda":  # [FIX-MEM] true per-fold training peak
        import gc; gc.collect(); torch.cuda.empty_cache(); torch.cuda.reset_peak_memory_stats(device)  # [FIX-LEAK]
    if SCALE_FEATURES:  # [FIX] z-score features with train-portion stats only
        _sc = StandardScaler().fit(features_raw[train_idx])
        features_raw = _sc.transform(features_raw).astype(np.float32)

    y_train = scores[train_idx]
    y_val   = scores[val_idx]

    train_ds = EssayDataset([texts[i] for i in train_idx],
                             features_raw[train_idx], y_train)
    val_ds   = EssayDataset([texts[i] for i in val_idx],
                             features_raw[val_idx],   y_val)
    train_dl = DataLoader(train_ds, batch_size=cfg["batch_size"],
                          shuffle=True,  num_workers=0, collate_fn=collate_fn)
    val_dl   = DataLoader(val_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0, collate_fn=collate_fn)
    # [FIX-LEAKAGE] early-stopping set (NOT the test fold)
    y_es     = scores[es_idx]
    es_ds   = EssayDataset([texts[i] for i in es_idx],
                             features_raw[es_idx],   y_es)
    es_dl   = DataLoader(es_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0, collate_fn=collate_fn)

    model = IndoSBERTRegressor(cfg["model_name"], features_raw.shape[1],
                                device).to(device)

    pipeline_params = (list(model.sbert[0].parameters()) +
                       list(model.sbert[2].parameters()))
    head_params     = (list(model.fc1.parameters()) +
                       list(model.fc2.parameters()))

    optimizer = torch.optim.AdamW([
        {"params": pipeline_params, "lr": cfg["lr_encoder"]},
        {"params": head_params,     "lr": cfg["lr_head"]},
    ], weight_decay=cfg["weight_decay"])

    total_steps = len(train_dl) * cfg["max_epochs"]
    scheduler   = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=cfg["warmup_steps"],
        num_training_steps=total_steps)
    criterion = nn.MSELoss()

    best_val_loss, best_model_state, patience_counter = float("inf"), None, 0
    train_loss_history, val_loss_history = [], []
    val_qwk_history,   val_mse_history   = [], []
    n_epochs_actual = 0

    t_start = time.perf_counter()
    for epoch in range(cfg["max_epochs"]):
        n_epochs_actual = epoch + 1
        train_loss = train_one_epoch(model, train_dl, optimizer,
                                     scheduler, criterion, device)
        val_loss, val_mse, val_qwk, _, _ = evaluate(
            model, es_dl, device, cfg["score_min"], cfg["score_max"])

        train_loss_history.append(train_loss)
        val_loss_history.append(val_loss)
        val_qwk_history.append(val_qwk)
        val_mse_history.append(val_mse)

        print(f"  Epoch {epoch+1:02d}/{cfg['max_epochs']} | "
              f"train={train_loss:.4f} | val={val_loss:.4f} | "
              f"QWK={val_qwk:.4f} | MSE={val_mse:.4f}")

        if val_loss < best_val_loss:
            best_val_loss    = val_loss
            best_model_state = {k: v.clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= cfg["patience"]:
                print(f"  Early stopping pada epoch {epoch+1}")
                break

    training_time = time.perf_counter() - t_start

    model_path = os.path.join(cfg["output_dir"], f"model_fold_{fold_idx+1}.pth")
    torch.save(best_model_state, model_path)
    model.load_state_dict(best_model_state)

    # --- Inference time ---
    model.eval()
    sample_batch = next(iter(val_dl))
    N_INFER = min(100, len(val_ds))
    t0 = time.perf_counter()
    with torch.no_grad():
        for _ in range(N_INFER):
            model([sample_batch["texts"][0]],
                  sample_batch["features"][:1].to(device))
    inference_ms = (time.perf_counter() - t0) / N_INFER * 1000

    # --- Memory tracking via pynvml (GPU only) ---
    if device.type == "cuda":
        pynvml.nvmlInit()
        handle   = pynvml.nvmlDeviceGetHandleByIndex(device.index or 0)
        mem_info = pynvml.nvmlDeviceGetMemoryInfo(handle)
        mem_mb   = torch.cuda.max_memory_allocated(device) / (1024 ** 2)  # [FIX-MEM] was whole-GPU "used"
        mem_device = "cuda"
        pynvml.nvmlShutdown()
    else:
        mem_mb     = 0.0
        mem_device = "cpu (not tracked)"

    _, final_mse, final_qwk, preds, true = evaluate(
        model, val_dl, device, cfg["score_min"], cfg["score_max"])

    fold_preds_df = pd.DataFrame({
        "sample_idx"  : val_idx,
        "question_id" : question_ids[val_idx],
        "true_score"  : true,
        "pred_score"  : preds,
        "residual"    : preds - true,
        "abs_error"   : np.abs(preds - true),
        "fold"        : fold_idx + 1,
    })

    print(f"\n  → Fold {fold_idx+1}: QWK={final_qwk:.4f} | MSE={final_mse:.4f} | "
          f"time={training_time:.1f}s | inf={inference_ms:.2f}ms | mem={mem_mb:.1f}MB")

    return {
        "fold"            : fold_idx + 1,
        "qwk"             : final_qwk,
        "mse"             : final_mse,
        "training_time_s" : training_time,
        "inference_ms"    : inference_ms,
        "memory_mb"       : mem_mb,
        "memory_device"   : mem_device,
        "best_val_loss"   : best_val_loss,
        "n_epochs_actual" : n_epochs_actual,
        "model_path"      : model_path,
        "val_idx"         : val_idx.tolist(),
        "train_idx"       : train_idx.tolist(),
        "history": {
            "train_loss": train_loss_history,
            "val_loss"  : val_loss_history,
            "val_qwk"   : val_qwk_history,
            "val_mse"   : val_mse_history,
        },
    }, fold_preds_df


# ══════════════════════════════════════════════════════════
#  MAIN
# ══════════════════════════════════════════════════════════
def main():
    set_seed(CFG["seed"])

    print("\n[1/4] Membaca dataset...")
    texts  = joblib.load(CFG["essay"])
    scores = joblib.load(CFG["score"])

    if os.path.exists(CFG["question_ids"]):
        question_ids = joblib.load(CFG["question_ids"])
    else:
        n_per_q = len(texts) // 10
        question_ids = np.repeat(np.arange(1, 11), n_per_q).astype(int)
        if len(question_ids) < len(texts):
            question_ids = np.append(question_ids,
                np.full(len(texts) - len(question_ids), 10, dtype=int))
        joblib.dump(question_ids, CFG["question_ids"])
        print(f"  question_ids dibuat ({n_per_q} esai/soal, asumsi berblok)")

    CFG["score_min"] = int(scores.min())
    CFG["score_max"] = int(scores.max())
    print(f"  Sampel: {len(texts)} | Skor: {CFG['score_min']}–{CFG['score_max']} | "
          f"Mean: {scores.mean():.2f}±{scores.std():.2f}")
    print(f"  PENTING: do_lower_case=False — teks tidak di-lowercase")

    print("\n[2/4] Mengekstrak fitur linguistik...")
    fe_list = []
    for i, text in enumerate(texts):
        if (i+1) % 100 == 0 or i == 0:
            print(f"  [{i+1}/{len(texts)}]")
        fe_list.append(extract_linguistic_features(text))
    features_raw = np.array([[d[k] for k in FEATURE_NAMES] for d in fe_list],
                              dtype=np.float32)
    print(f"  Shape: {features_raw.shape}")

    print(f"\n[3/4] Memulai {CFG['n_folds']}-fold CV...")
    kf = KFold(n_splits=CFG["n_folds"], shuffle=True, random_state=CFG["seed"])
    all_results, all_preds_dfs = [], []

    for fold_idx, (train_idx, val_idx) in enumerate(kf.split(texts)):

        train_idx, es_idx = inner_split(train_idx, scores, fold_idx)  # [FIX-LEAKAGE]
        try:
            result, preds_df = run_fold(
                fold_idx, train_idx, val_idx,
                texts, features_raw, scores, question_ids,
                CFG, DEVICE, es_idx=es_idx)
            all_results.append(result)
            all_preds_dfs.append(preds_df)
        except Exception as e:
            print(f"  [ERROR] Fold {fold_idx+1}: {e}")
            traceback.print_exc()

    print(f"\n[4/4] Menyimpan hasil...")
    all_preds_df = pd.concat(all_preds_dfs, ignore_index=True)
    overall_qwk  = quadratic_weighted_kappa(
        all_preds_df["true_score"].values, all_preds_df["pred_score"].values,
        CFG["score_min"], CFG["score_max"])
    overall_mse  = mean_squared_error(
        all_preds_df["true_score"].values, all_preds_df["pred_score"].values)

    score_dist = {}
    for q in range(1, 11):
        mask = question_ids == q
        if mask.sum() > 0:
            score_dist[int(q)] = {
                "mean"  : float(scores[mask].mean()),
                "std"   : float(scores[mask].std()),
                "n"     : int(mask.sum()),
                "counts": {int(k): int(v) for k, v in
                           zip(*np.unique(scores[mask].astype(int),
                                          return_counts=True))},
            }

    experiment_data = {
        "config"            : CFG,
        "fold_results"      : all_results,
        "fold_predictions"  : all_preds_df,
        "overall_qwk"       : overall_qwk,
        "overall_mse"       : overall_mse,
        "features_raw"      : features_raw,
        "feature_names"     : FEATURE_NAMES,
        "scores"            : scores,
        "question_ids"      : question_ids,
        "score_distribution": score_dist,
        "model_name"        : CFG["model_name"],
        "paradigm"          : "fine-tuning",
        "n_samples"         : len(texts),
        "timestamp"         : datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    }

    joblib_path = os.path.join(CFG["output_dir"], "experiment_results.joblib")
    joblib.dump(experiment_data, joblib_path)

    results_df     = pd.DataFrame(all_results)
    results_df_csv = results_df.drop(columns=["history","val_idx","train_idx"],
                                      errors="ignore")
    results_df_csv.to_csv(
        os.path.join(CFG["output_dir"], "cv_results_indosbert_ft.csv"), index=False)
    all_preds_df.to_csv(
        os.path.join(CFG["output_dir"], "all_predictions_indosbert_ft.csv"), index=False)

    print(f"\n{'='*55}")
    print("  RINGKASAN — IndoSBERT-large Fine-tuning + FE")
    print(f"{'='*55}")
    print(results_df_csv.to_string(index=False))
    print(f"\n  Rata-rata ± Std:")
    for m in ["qwk","mse","training_time_s","inference_ms","memory_mb"]:
        print(f"    {m:<20s}: {results_df[m].mean():.4f} ± {results_df[m].std():.4f}")
    print(f"\n  QWK keseluruhan : {overall_qwk:.4f}")
    print(f"  MSE keseluruhan : {overall_mse:.4f}")
    print(f"\n  Disimpan di     : {joblib_path}")

    return experiment_data


if __name__ == "__main__":
    main()
```

```python
"""
=============================================================
  AES — Paradigma A: IndoBERT Fine-tuning + FE Augmentation
  Dataset  : preprocessed_texts.pkl, scores.pkl
  Model    : indobenchmark/indobert-base-p2
  Metrik   : QWK, MSE
  Evaluasi : 10-Fold Cross-Validation
  Referensi: Devlin et al. (2019), Mosbach et al. (2021),
             Dodge et al. (2020), Iqbal et al. (2024)
=============================================================
"""

import os, time, traceback, random
from datetime import datetime

import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import pynvml
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel, get_linear_schedule_with_warmup
from sklearn.model_selection import KFold
# ── [FIX-LEAKAGE] inner early-stopping split ────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

ES_FRAC        = 0.10   # share of the outer-train fold used for early stopping
SCALE_FEATURES = True   # FA-FT only: z-score linguistic features (fit on train only)

def inner_split(train_idx, y, fold_idx, frac=ES_FRAC, seed=42):
    """Split the outer-train fold into (train, early-stop) sets.
    Stratified by score band (scores <=5 merged) so rare low scores appear in both.
    Same seed in every notebook -> identical splits for FA-FT and FA-ML."""
    bands = np.clip(np.asarray(y)[train_idx].astype(int), 5, 10)
    tr, es = train_test_split(train_idx, test_size=frac,
                              random_state=seed + fold_idx, stratify=bands)
    return np.sort(tr), np.sort(es)
# ────────────────────────────────────────────────────────────────────────
from sklearn.metrics import mean_squared_error
from nltk.tokenize import word_tokenize, sent_tokenize
from nlp_id.postag import PosTag


# ══════════════════════════════════════════════════════════
#  KONFIGURASI
# ══════════════════════════════════════════════════════════
CFG = {
    "essay"        : f"{ESSAY_PATH}preprocessed_texts.pkl",
    "score"        : f"{SCORE_PATH}scores.pkl",
    "question_ids" : f"{SCORE_PATH}question_ids.pkl",
    "model_name"   : "indobenchmark/indobert-base-p2",
    "max_len"      : 512,
    "lr_encoder"   : 2e-5,
    "lr_head"      : 1e-3,
    "weight_decay" : 0.01,
    "warmup_ratio" : 0.10,
    "max_epochs"   : 30,
    "batch_size"   : 64,
    "patience"     : 5,
    "dropout"      : 0.1,
    "n_folds"      : 10,
    "seed"         : 42,
    "output_dir"   : f"{OUTPUT_PATH}output_indobert_ft",
    "score_min"    : 2,
    "score_max"    : 10,
}

os.makedirs(CFG["output_dir"], exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {DEVICE}")
print(f"Model  : {CFG['model_name']}")


# ══════════════════════════════════════════════════════════
#  REPRODUCIBILITY
# ══════════════════════════════════════════════════════════
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ══════════════════════════════════════════════════════════
#  LINGUISTIC FEATURE EXTRACTION
#  6 fitur — dihitung dari teks preprocessed yang mempertahankan
#  tanda baca (bukan dari raw text terpisah).
#
#  Fitur:
#  1. ttr              — lexical diversity (unique/total tokens)
#  2. word_count       — panjang respons, proxy cakupan konten
#  3. noun_count       — kepadatan konten faktual (NN, NNP)
#  4. verb_count       — elaborasi naratif/kausal (VB)
#  5. sentence_count   — kompleksitas sintaksis (proxy)
#  6. avg_word_length  — kompleksitas morfologi (proxy, Indonesian-specific)
# ══════════════════════════════════════════════════════════
postag = PosTag()

POS_NOUN = {'NN', 'NNP'}
POS_VERB = {'VB'}

FEATURE_NAMES = [
    'ttr',
    'word_count',
    'noun_count',
    'verb_count',
    'sentence_count',
    'avg_word_length',
]


def extract_linguistic_features(text: str) -> dict:
    if not text:
        return {f: 0.0 for f in FEATURE_NAMES}

    words       = word_tokenize(text)
    word_tokens = [w for w in words if w.isalnum()]
    n_words     = len(word_tokens)

    if n_words == 0:
        return {f: 0.0 for f in FEATURE_NAMES}

    ttr             = len(set(word_tokens)) / n_words
    word_count      = float(n_words)
    pos_tags        = postag.get_pos_tag(text)
    noun_count      = sum(1 for _, pos in pos_tags if pos in POS_NOUN)
    verb_count      = sum(1 for _, pos in pos_tags if pos in POS_VERB)
    sentences       = sent_tokenize(text)
    sentence_count  = float(max(len(sentences), 1))
    avg_word_length = sum(len(w) for w in word_tokens) / n_words

    return {
        'ttr'            : ttr,
        'word_count'     : word_count,
        'noun_count'     : float(noun_count),
        'verb_count'     : float(verb_count),
        'sentence_count' : sentence_count,
        'avg_word_length': avg_word_length,
    }


# ══════════════════════════════════════════════════════════
#  QWK
# ══════════════════════════════════════════════════════════
def quadratic_weighted_kappa(y_true, y_pred, min_rating, max_rating):
    y_true = np.array(y_true, dtype=int)
    y_pred = np.round(np.clip(y_pred, min_rating, max_rating)).astype(int)
    num_ratings = max_rating - min_rating + 1
    conf_mat    = np.zeros((num_ratings, num_ratings), dtype=int)
    for a, p in zip(y_true, y_pred):
        conf_mat[a - min_rating][p - min_rating] += 1
    num_scored = len(y_true)
    hist_true  = np.bincount(y_true - min_rating, minlength=num_ratings)
    hist_pred  = np.bincount(y_pred - min_rating, minlength=num_ratings)
    expected   = np.outer(hist_true, hist_pred) / num_scored
    weights    = np.array([[(i - j) ** 2 / (num_ratings - 1) ** 2
                             for j in range(num_ratings)]
                            for i in range(num_ratings)])
    denom = np.sum(weights * expected)
    return 0.0 if denom == 0 else 1.0 - np.sum(weights * conf_mat) / denom


# ══════════════════════════════════════════════════════════
#  DATASET
# ══════════════════════════════════════════════════════════
class EssayDataset(Dataset):
    def __init__(self, texts, features, scores, tokenizer, max_len):
        self.texts     = texts
        self.features  = features
        self.scores    = scores
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self): return len(self.texts)

    def __getitem__(self, idx):
        enc = self.tokenizer(self.texts[idx], max_length=self.max_len,
                             padding="max_length", truncation=True,
                             return_tensors="pt")
        return {
            "input_ids"     : enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "features"      : torch.tensor(self.features[idx], dtype=torch.float32),
            "score"         : torch.tensor(self.scores[idx],   dtype=torch.float32),
        }


# ══════════════════════════════════════════════════════════
#  MODEL
# ══════════════════════════════════════════════════════════
class IndoBERTRegressor(nn.Module):
    def __init__(self, model_name, n_feat, dropout=0.1):
        super().__init__()
        self.bert    = AutoModel.from_pretrained(model_name)
        hidden_size  = self.bert.config.hidden_size  # 768
        self.dropout = nn.Dropout(dropout)
        self.fc1     = nn.Linear(hidden_size + n_feat, 256)
        self.relu    = nn.ReLU()
        self.fc2     = nn.Linear(256, 1)

    def forward(self, input_ids, attention_mask, features):
        out    = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls    = self.dropout(out.last_hidden_state[:, 0, :])
        concat = torch.cat([cls, features], dim=1)
        return self.fc2(self.dropout(self.relu(self.fc1(concat)))).squeeze(-1)


# ══════════════════════════════════════════════════════════
#  TRAINING & EVALUATION
# ══════════════════════════════════════════════════════════
def train_one_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    for batch in loader:
        optimizer.zero_grad()
        preds = model(batch["input_ids"].to(device),
                      batch["attention_mask"].to(device),
                      batch["features"].to(device))
        loss  = criterion(preds, batch["score"].to(device))
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
    return total_loss / len(loader)


def evaluate(model, loader, device, score_min, score_max):
    model.eval()
    all_preds, all_true, total_loss = [], [], 0.0
    criterion = nn.MSELoss()
    with torch.no_grad():
        for batch in loader:
            preds = model(batch["input_ids"].to(device),
                          batch["attention_mask"].to(device),
                          batch["features"].to(device))
            total_loss += criterion(preds, batch["score"].to(device)).item()
            all_preds.extend(preds.cpu().numpy().tolist())
            all_true.extend(batch["score"].cpu().numpy().tolist())
    all_preds = np.array(all_preds)
    all_true  = np.array(all_true)
    return (total_loss / len(loader),
            mean_squared_error(all_true, all_preds),
            quadratic_weighted_kappa(all_true, all_preds, score_min, score_max),
            all_preds, all_true)


# ══════════════════════════════════════════════════════════
#  RUN FOLD
# ══════════════════════════════════════════════════════════
def run_fold(fold_idx, train_idx, val_idx, texts, features_raw,
             scores, question_ids, tokenizer, cfg, device, es_idx=None):

    print(f"\n{'='*55}\n  FOLD {fold_idx+1}/{cfg['n_folds']}\n{'='*55}")
    set_seed(cfg["seed"] + fold_idx)
    if device.type == "cuda":  # [FIX-MEM] true per-fold training peak
        import gc; gc.collect(); torch.cuda.empty_cache(); torch.cuda.reset_peak_memory_stats(device)  # [FIX-LEAK]
    if SCALE_FEATURES:  # [FIX] z-score features with train-portion stats only
        _sc = StandardScaler().fit(features_raw[train_idx])
        features_raw = _sc.transform(features_raw).astype(np.float32)

    y_train = scores[train_idx]
    y_val   = scores[val_idx]

    train_ds = EssayDataset([texts[i] for i in train_idx],
                             features_raw[train_idx], y_train,
                             tokenizer, cfg["max_len"])
    val_ds   = EssayDataset([texts[i] for i in val_idx],
                             features_raw[val_idx],   y_val,
                             tokenizer, cfg["max_len"])
    train_dl = DataLoader(train_ds, batch_size=cfg["batch_size"],
                          shuffle=True,  num_workers=0)
    val_dl   = DataLoader(val_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0)
    # [FIX-LEAKAGE] early-stopping set (NOT the test fold)
    y_es     = scores[es_idx]
    es_ds   = EssayDataset([texts[i] for i in es_idx],
                             features_raw[es_idx],   y_es,
                             tokenizer, cfg["max_len"])
    es_dl   = DataLoader(es_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0)

    model = IndoBERTRegressor(cfg["model_name"], features_raw.shape[1],
                               cfg["dropout"]).to(device)

    optimizer = torch.optim.AdamW([
        {"params": list(model.bert.parameters()), "lr": cfg["lr_encoder"]},
        {"params": list(model.fc1.parameters()) + list(model.fc2.parameters()),
         "lr": cfg["lr_head"]},
    ], weight_decay=cfg["weight_decay"])

    total_steps  = len(train_dl) * cfg["max_epochs"]
    warmup_steps = int(total_steps * cfg["warmup_ratio"])
    scheduler    = get_linear_schedule_with_warmup(
        optimizer, num_warmup_steps=warmup_steps,
        num_training_steps=total_steps)
    criterion = nn.MSELoss()

    best_val_loss, best_model_state, patience_counter = float("inf"), None, 0
    train_loss_history, val_loss_history = [], []
    val_qwk_history,   val_mse_history   = [], []
    n_epochs_actual = 0

    t_start = time.perf_counter()
    for epoch in range(cfg["max_epochs"]):
        n_epochs_actual = epoch + 1
        train_loss = train_one_epoch(model, train_dl, optimizer,
                                     scheduler, criterion, device)
        val_loss, val_mse, val_qwk, _, _ = evaluate(
            model, es_dl, device, cfg["score_min"], cfg["score_max"])

        train_loss_history.append(train_loss)
        val_loss_history.append(val_loss)
        val_qwk_history.append(val_qwk)
        val_mse_history.append(val_mse)

        print(f"  Epoch {epoch+1:02d}/{cfg['max_epochs']} | "
              f"train={train_loss:.4f} | val={val_loss:.4f} | "
              f"QWK={val_qwk:.4f} | MSE={val_mse:.4f}")

        if val_loss < best_val_loss:
            best_val_loss    = val_loss
            best_model_state = {k: v.clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= cfg["patience"]:
                print(f"  Early stopping pada epoch {epoch+1}")
                break

    training_time = time.perf_counter() - t_start

    # Load best state untuk evaluasi akhir (tanpa simpan ke disk)
    model.load_state_dict(best_model_state)

    # --- Inference time ---
    model.eval()
    sample_batch = next(iter(val_dl))
    N_INFER = min(100, len(val_ds))
    t0 = time.perf_counter()
    with torch.no_grad():
        for _ in range(N_INFER):
            model(sample_batch["input_ids"][:1].to(device),
                  sample_batch["attention_mask"][:1].to(device),
                  sample_batch["features"][:1].to(device))
    inference_ms = (time.perf_counter() - t0) / N_INFER * 1000

    # --- Memory tracking via pynvml (GPU only) ---
    if device.type == "cuda":
        pynvml.nvmlInit()
        handle   = pynvml.nvmlDeviceGetHandleByIndex(device.index or 0)
        mem_info = pynvml.nvmlDeviceGetMemoryInfo(handle)
        mem_mb   = torch.cuda.max_memory_allocated(device) / (1024 ** 2)  # [FIX-MEM] was whole-GPU "used"
        mem_device = "cuda"
        pynvml.nvmlShutdown()
    else:
        mem_mb     = 0.0
        mem_device = "cpu (not tracked)"

    # --- Evaluasi akhir ---
    _, final_mse, final_qwk, preds, true = evaluate(
        model, val_dl, device, cfg["score_min"], cfg["score_max"])

    fold_preds_df = pd.DataFrame({
        "sample_idx"  : val_idx,
        "question_id" : question_ids[val_idx],
        "true_score"  : true,
        "pred_score"  : preds,
        "residual"    : preds - true,
        "abs_error"   : np.abs(preds - true),
        "fold"        : fold_idx + 1,
    })

    print(f"\n  → Fold {fold_idx+1}: QWK={final_qwk:.4f} | MSE={final_mse:.4f} | "
          f"time={training_time:.1f}s | inf={inference_ms:.2f}ms | mem={mem_mb:.1f}MB")

    return {
        "fold"            : fold_idx + 1,
        "qwk"             : final_qwk,
        "mse"             : final_mse,
        "training_time_s" : training_time,
        "inference_ms"    : inference_ms,
        "memory_mb"       : mem_mb,
        "memory_device"   : mem_device,
        "best_val_loss"   : best_val_loss,
        "n_epochs_actual" : n_epochs_actual,
        "val_idx"         : val_idx.tolist(),
        "train_idx"       : train_idx.tolist(),
        "history": {
            "train_loss": train_loss_history,
            "val_loss"  : val_loss_history,
            "val_qwk"   : val_qwk_history,
            "val_mse"   : val_mse_history,
        },
    }, fold_preds_df


# ══════════════════════════════════════════════════════════
#  MAIN
# ══════════════════════════════════════════════════════════
def main():
    set_seed(CFG["seed"])

    print("\n[1/5] Membaca dataset...")
    texts  = joblib.load(CFG["essay"])
    scores = joblib.load(CFG["score"])

    if os.path.exists(CFG["question_ids"]):
        question_ids = joblib.load(CFG["question_ids"])
    else:
        n_per_q = len(texts) // 10
        question_ids = np.repeat(np.arange(1, 11), n_per_q).astype(int)
        if len(question_ids) < len(texts):
            question_ids = np.append(question_ids,
                np.full(len(texts) - len(question_ids), 10, dtype=int))
        joblib.dump(question_ids, CFG["question_ids"])
        print(f"  question_ids dibuat dan disimpan (asumsi {n_per_q} esai/soal)")

    CFG["score_min"] = int(scores.min())
    CFG["score_max"] = int(scores.max())
    print(f"  Sampel: {len(texts)} | Skor: {CFG['score_min']}–{CFG['score_max']} | "
          f"Mean: {scores.mean():.2f}±{scores.std():.2f}")

    print("\n[2/5] Mengekstrak fitur linguistik...")
    fe_list = []
    for i, text in enumerate(texts):
        if (i+1) % 100 == 0 or i == 0:
            print(f"  [{i+1}/{len(texts)}]")
        fe_list.append(extract_linguistic_features(text))
    features_raw = np.array([[d[k] for k in FEATURE_NAMES] for d in fe_list],
                              dtype=np.float32)
    print(f"  Shape: {features_raw.shape} | Fitur: {FEATURE_NAMES}")

    print(f"\n[3/5] Memuat tokenizer: {CFG['model_name']}...")
    tokenizer = AutoTokenizer.from_pretrained(CFG["model_name"])

    print(f"\n[4/5] Memulai {CFG['n_folds']}-fold CV...")
    kf = KFold(n_splits=CFG["n_folds"], shuffle=True, random_state=CFG["seed"])
    all_results, all_preds_dfs = [], []

    for fold_idx, (train_idx, val_idx) in enumerate(kf.split(texts)):

        train_idx, es_idx = inner_split(train_idx, scores, fold_idx)  # [FIX-LEAKAGE]
        try:
            result, preds_df = run_fold(
                fold_idx, train_idx, val_idx,
                texts, features_raw, scores, question_ids,
                tokenizer, CFG, DEVICE, es_idx=es_idx)
            all_results.append(result)
            all_preds_dfs.append(preds_df)
        except Exception as e:
            print(f"  [ERROR] Fold {fold_idx+1}: {e}")
            traceback.print_exc()

    print(f"\n[5/5] Menyimpan hasil...")
    all_preds_df = pd.concat(all_preds_dfs, ignore_index=True)

    overall_qwk = quadratic_weighted_kappa(
        all_preds_df["true_score"].values,
        all_preds_df["pred_score"].values,
        CFG["score_min"], CFG["score_max"])
    overall_mse = mean_squared_error(
        all_preds_df["true_score"].values,
        all_preds_df["pred_score"].values)

    score_dist = {}
    for q in range(1, 11):
        mask = question_ids == q
        if mask.sum() > 0:
            score_dist[int(q)] = {
                "mean"  : float(scores[mask].mean()),
                "std"   : float(scores[mask].std()),
                "n"     : int(mask.sum()),
                "counts": {int(k): int(v) for k, v in
                           zip(*np.unique(scores[mask].astype(int),
                                          return_counts=True))},
            }

    experiment_data = {
        "config"            : CFG,
        "fold_results"      : all_results,
        "fold_predictions"  : all_preds_df,
        "overall_qwk"       : overall_qwk,
        "overall_mse"       : overall_mse,
        "features_raw"      : features_raw,
        "feature_names"     : FEATURE_NAMES,
        "scores"            : scores,
        "question_ids"      : question_ids,
        "score_distribution": score_dist,
        "model_name"        : CFG["model_name"],
        "paradigm"          : "fine-tuning",
        "n_samples"         : len(texts),
        "timestamp"         : datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    }

    joblib_path = os.path.join(CFG["output_dir"], "experiment_results.joblib")
    joblib.dump(experiment_data, joblib_path)

    results_df     = pd.DataFrame(all_results)
    results_df_csv = results_df.drop(columns=["history","val_idx","train_idx"],
                                      errors="ignore")
    results_df_csv.to_csv(
        os.path.join(CFG["output_dir"], "cv_results_indobert_ft.csv"), index=False)
    all_preds_df.to_csv(
        os.path.join(CFG["output_dir"], "all_predictions_indobert_ft.csv"), index=False)

    print(f"\n{'='*55}")
    print("  RINGKASAN — IndoBERT Fine-tuning + FE")
    print(f"{'='*55}")
    print(results_df_csv.to_string(index=False))
    print(f"\n  Rata-rata ± Std:")
    for m in ["qwk","mse","training_time_s","inference_ms","memory_mb"]:
        print(f"    {m:<20s}: {results_df[m].mean():.4f} ± {results_df[m].std():.4f}")
    print(f"\n  QWK keseluruhan : {overall_qwk:.4f}")
    print(f"  MSE keseluruhan : {overall_mse:.4f}")
    print(f"\n  Disimpan di     : {joblib_path}")
    print(f"  Keys            : {list(experiment_data.keys())}")

    return experiment_data


if __name__ == "__main__":
    main()
```

```python
"""
=============================================================
  AES — Paradigma A: DistilBERT-Indo Fine-tuning + FE Augmentation
  Dataset  : preprocessed_texts.pkl, scores.pkl
  Model    : cahya/distilbert-base-indonesian
  Arsitektur: DistilBERT 6L (from scratch, Indo) → [CLS] → 768
  Metrik   : QWK, MSE
  Evaluasi : 10-Fold Cross-Validation (sama seperti IndoBERT)
=============================================================
"""

import os, time, traceback, random
from datetime import datetime

import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import pynvml
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel, get_linear_schedule_with_warmup
from sklearn.model_selection import KFold
# ── [FIX-LEAKAGE] inner early-stopping split ────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

ES_FRAC        = 0.10   # share of the outer-train fold used for early stopping
SCALE_FEATURES = True   # FA-FT only: z-score linguistic features (fit on train only)

def inner_split(train_idx, y, fold_idx, frac=ES_FRAC, seed=42):
    """Split the outer-train fold into (train, early-stop) sets.
    Stratified by score band (scores <=5 merged) so rare low scores appear in both.
    Same seed in every notebook -> identical splits for FA-FT and FA-ML."""
    bands = np.clip(np.asarray(y)[train_idx].astype(int), 5, 10)
    tr, es = train_test_split(train_idx, test_size=frac,
                              random_state=seed + fold_idx, stratify=bands)
    return np.sort(tr), np.sort(es)
# ────────────────────────────────────────────────────────────────────────
from sklearn.metrics import mean_squared_error
from nltk.tokenize import word_tokenize, sent_tokenize
from nlp_id.postag import PosTag


# ══════════════════════════════════════════════════════════
#  KONFIGURASI
# ══════════════════════════════════════════════════════════
CFG = {
    "essay"        : f"{ESSAY_PATH}preprocessed_texts.pkl",
    "score"        : f"{SCORE_PATH}scores.pkl",
    "question_ids" : f"{SCORE_PATH}question_ids.pkl",
    "model_name"   : "cahya/distilbert-base-indonesian",
    "max_len"      : 512,
    "lr_encoder"   : 3e-5,
    "lr_head"      : 1e-3,
    "weight_decay" : 0.01,
    "warmup_ratio" : 0.10,
    "max_epochs"   : 30,
    "batch_size"   : 64,
    "patience"     : 5,
    "dropout"      : 0.1,
    "n_folds"      : 10,
    "seed"         : 42,
    "output_dir"   : f"{OUTPUT_PATH}Hasil eksperiment/output_distilbert_ft",
    "score_min"    : 2,
    "score_max"    : 10,
}

os.makedirs(CFG["output_dir"], exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device : {DEVICE}")
print(f"Model  : {CFG['model_name']} (DistilBERT 6L, [CLS], hidden 768)")


# ══════════════════════════════════════════════════════════
#  REPRODUCIBILITY
# ══════════════════════════════════════════════════════════
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ══════════════════════════════════════════════════════════
#  LINGUISTIC FEATURE EXTRACTION
#  6 fitur — dihitung dari teks preprocessed yang mempertahankan
#  tanda baca (bukan dari raw text terpisah).
#
#  Fitur:
#  1. ttr              — lexical diversity (unique/total tokens)
#  2. word_count       — panjang respons, proxy cakupan konten
#  3. noun_count       — kepadatan konten faktual (NN, NNP)
#  4. verb_count       — elaborasi naratif/kausal (VB)
#  5. sentence_count   — kompleksitas sintaksis (proxy)
#  6. avg_word_length  — kompleksitas morfologi (proxy, Indonesian-specific)
# ══════════════════════════════════════════════════════════
postag = PosTag()

POS_NOUN = {'NN', 'NNP'}
POS_VERB = {'VB'}

FEATURE_NAMES = [
    'ttr',
    'word_count',
    'noun_count',
    'verb_count',
    'sentence_count',
    'avg_word_length',
]


def extract_linguistic_features(text: str) -> dict:
    if not text:
        return {f: 0.0 for f in FEATURE_NAMES}

    words       = word_tokenize(text)
    word_tokens = [w for w in words if w.isalnum()]
    n_words     = len(word_tokens)

    if n_words == 0:
        return {f: 0.0 for f in FEATURE_NAMES}

    ttr             = len(set(word_tokens)) / n_words
    word_count      = float(n_words)
    pos_tags        = postag.get_pos_tag(text)
    noun_count      = sum(1 for _, pos in pos_tags if pos in POS_NOUN)
    verb_count      = sum(1 for _, pos in pos_tags if pos in POS_VERB)
    sentences       = sent_tokenize(text)
    sentence_count  = float(max(len(sentences), 1))
    avg_word_length = sum(len(w) for w in word_tokens) / n_words

    return {
        'ttr'            : ttr,
        'word_count'     : word_count,
        'noun_count'     : float(noun_count),
        'verb_count'     : float(verb_count),
        'sentence_count' : sentence_count,
        'avg_word_length': avg_word_length,
    }


# ══════════════════════════════════════════════════════════
#  QWK
# ══════════════════════════════════════════════════════════
def quadratic_weighted_kappa(y_true, y_pred, min_rating, max_rating):
    y_true = np.array(y_true, dtype=int)
    y_pred = np.round(np.clip(y_pred, min_rating, max_rating)).astype(int)
    num_ratings = max_rating - min_rating + 1
    conf_mat    = np.zeros((num_ratings, num_ratings), dtype=int)
    for a, p in zip(y_true, y_pred):
        conf_mat[a - min_rating][p - min_rating] += 1
    num_scored = len(y_true)
    hist_true  = np.bincount(y_true - min_rating, minlength=num_ratings)
    hist_pred  = np.bincount(y_pred - min_rating, minlength=num_ratings)
    expected   = np.outer(hist_true, hist_pred) / num_scored
    weights    = np.array([[(i - j) ** 2 / (num_ratings - 1) ** 2
                             for j in range(num_ratings)]
                            for i in range(num_ratings)])
    denom = np.sum(weights * expected)
    return 0.0 if denom == 0 else 1.0 - np.sum(weights * conf_mat) / denom


# ══════════════════════════════════════════════════════════
#  DATASET
# ══════════════════════════════════════════════════════════
class EssayDataset(Dataset):
    def __init__(self, texts, features, scores, tokenizer, max_len):
        self.texts     = texts
        self.features  = features
        self.scores    = scores
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self): return len(self.texts)

    def __getitem__(self, idx):
        enc = self.tokenizer(self.texts[idx], max_length=self.max_len,
                             padding="max_length", truncation=True,
                             return_tensors="pt")
        return {
            "input_ids"     : enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "features"      : torch.tensor(self.features[idx], dtype=torch.float32),
            "score"         : torch.tensor(self.scores[idx],   dtype=torch.float32),
        }


# ══════════════════════════════════════════════════════════
#  MODEL — DistilBERTRegressor
# ══════════════════════════════════════════════════════════
class DistilBERTRegressor(nn.Module):
    def __init__(self, model_name, n_feat, dropout=0.1):
        super().__init__()
        self.bert    = AutoModel.from_pretrained(model_name)
        hidden_size  = self.bert.config.hidden_size  # 768
        self.dropout = nn.Dropout(dropout)
        self.fc1     = nn.Linear(hidden_size + n_feat, 256)
        self.relu    = nn.ReLU()
        self.fc2     = nn.Linear(256, 1)

    def forward(self, input_ids, attention_mask, features):
        out    = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls    = self.dropout(out.last_hidden_state[:, 0, :])
        concat = torch.cat([cls, features], dim=1)
        return self.fc2(self.dropout(self.relu(self.fc1(concat)))).squeeze(-1)


# ══════════════════════════════════════════════════════════
#  TRAINING & EVALUATION
# ══════════════════════════════════════════════════════════
def train_one_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    for batch in loader:
        optimizer.zero_grad()
        preds = model(batch["input_ids"].to(device),
                      batch["attention_mask"].to(device),
                      batch["features"].to(device))
        loss  = criterion(preds, batch["score"].to(device))
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
    return total_loss / len(loader)


def evaluate(model, loader, device, score_min, score_max):
    model.eval()
    all_preds, all_true, total_loss = [], [], 0.0
    criterion = nn.MSELoss()
    with torch.no_grad():
        for batch in loader:
            preds = model(batch["input_ids"].to(device),
                          batch["attention_mask"].to(device),
                          batch["features"].to(device))
            total_loss += criterion(preds, batch["score"].to(device)).item()
            all_preds.extend(preds.cpu().numpy().tolist())
            all_true.extend(batch["score"].cpu().numpy().tolist())
    all_preds = np.array(all_preds)
    all_true  = np.array(all_true)
    return (total_loss / len(loader),
            mean_squared_error(all_true, all_preds),
            quadratic_weighted_kappa(all_true, all_preds, score_min, score_max),
            all_preds, all_true)


# ══════════════════════════════════════════════════════════
#  RUN FOLD
# ══════════════════════════════════════════════════════════
def run_fold(fold_idx, train_idx, val_idx, texts, features_raw,
             scores, question_ids, tokenizer, cfg, device, es_idx=None):

    print(f"\n{'='*55}\n  FOLD {fold_idx+1}/{cfg['n_folds']}\n{'='*55}")
    set_seed(cfg["seed"] + fold_idx)
    if device.type == "cuda":  # [FIX-MEM] true per-fold training peak
        import gc; gc.collect(); torch.cuda.empty_cache(); torch.cuda.reset_peak_memory_stats(device)  # [FIX-LEAK]
    if SCALE_FEATURES:  # [FIX] z-score features with train-portion stats only
        _sc = StandardScaler().fit(features_raw[train_idx])
        features_raw = _sc.transform(features_raw).astype(np.float32)

    y_train = scores[train_idx]
    y_val   = scores[val_idx]

    train_ds = EssayDataset([texts[i] for i in train_idx],
                             features_raw[train_idx], y_train,
                             tokenizer, cfg["max_len"])
    val_ds   = EssayDataset([texts[i] for i in val_idx],
                             features_raw[val_idx],   y_val,
                             tokenizer, cfg["max_len"])
    train_dl = DataLoader(train_ds, batch_size=cfg["batch_size"],
                          shuffle=True,  num_workers=0)
    val_dl   = DataLoader(val_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0)
    # [FIX-LEAKAGE] early-stopping set (NOT the test fold)
    y_es     = scores[es_idx]
    es_ds   = EssayDataset([texts[i] for i in es_idx],
                             features_raw[es_idx],   y_es,
                             tokenizer, cfg["max_len"])
    es_dl   = DataLoader(es_ds,   batch_size=cfg["batch_size"],
                          shuffle=False, num_workers=0)

    model = DistilBERTRegressor(cfg["model_name"], features_raw.shape[1],
                                 cfg["dropout"]).to(device)

    optimizer = torch.optim.AdamW([
        {"params": list(model.bert.parameters()), "lr": cfg["lr_encoder"]},
        {"params": list(model.fc1.parameters()) + list(model.fc2.parameters()),
         "lr": cfg["lr_head"]},
    ], weight_decay=cfg["weight_decay"])

    total_steps  = len(train_dl) * cfg["max_epochs"]
    warmup_steps = int(total_steps * cfg["warmup_ratio"])
    scheduler    = get_linear_schedule_with_warmup(
        optimizer, num_warmup_steps=warmup_steps,
        num_training_steps=total_steps)
    criterion = nn.MSELoss()

    best_val_loss, best_model_state, patience_counter = float("inf"), None, 0
    train_loss_history, val_loss_history = [], []
    val_qwk_history,   val_mse_history   = [], []
    n_epochs_actual = 0

    t_start = time.perf_counter()
    for epoch in range(cfg["max_epochs"]):
        n_epochs_actual = epoch + 1
        train_loss = train_one_epoch(model, train_dl, optimizer,
                                     scheduler, criterion, device)
        val_loss, val_mse, val_qwk, _, _ = evaluate(
            model, es_dl, device, cfg["score_min"], cfg["score_max"])

        train_loss_history.append(train_loss)
        val_loss_history.append(val_loss)
        val_qwk_history.append(val_qwk)
        val_mse_history.append(val_mse)

        print(f"  Epoch {epoch+1:02d}/{cfg['max_epochs']} | "
              f"train={train_loss:.4f} | val={val_loss:.4f} | "
              f"QWK={val_qwk:.4f} | MSE={val_mse:.4f}")

        if val_loss < best_val_loss:
            best_val_loss    = val_loss
            best_model_state = {k: v.clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= cfg["patience"]:
                print(f"  Early stopping pada epoch {epoch+1}")
                break

    training_time = time.perf_counter() - t_start

    # Load best state (tanpa simpan ke disk, seperti IndoBERT)
    model.load_state_dict(best_model_state)

    # --- Inference time ---
    model.eval()
    sample_batch = next(iter(val_dl))
    N_INFER = min(100, len(val_ds))
    t0 = time.perf_counter()
    with torch.no_grad():
        for _ in range(N_INFER):
            model(sample_batch["input_ids"][:1].to(device),
                  sample_batch["attention_mask"][:1].to(device),
                  sample_batch["features"][:1].to(device))
    inference_ms = (time.perf_counter() - t0) / N_INFER * 1000

    # --- Memory tracking via pynvml (GPU only) ---
    if device.type == "cuda":
        pynvml.nvmlInit()
        handle   = pynvml.nvmlDeviceGetHandleByIndex(device.index or 0)
        mem_info = pynvml.nvmlDeviceGetMemoryInfo(handle)
        mem_mb   = torch.cuda.max_memory_allocated(device) / (1024 ** 2)  # [FIX-MEM] was whole-GPU "used"
        mem_device = "cuda"
        pynvml.nvmlShutdown()
    else:
        mem_mb     = 0.0
        mem_device = "cpu (not tracked)"

    # --- Evaluasi akhir ---
    _, final_mse, final_qwk, preds, true = evaluate(
        model, val_dl, device, cfg["score_min"], cfg["score_max"])

    fold_preds_df = pd.DataFrame({
        "sample_idx"  : val_idx,
        "question_id" : question_ids[val_idx],
        "true_score"  : true,
        "pred_score"  : preds,
        "residual"    : preds - true,
        "abs_error"   : np.abs(preds - true),
        "fold"        : fold_idx + 1,
    })

    print(f"\n  → Fold {fold_idx+1}: QWK={final_qwk:.4f} | MSE={final_mse:.4f} | "
          f"time={training_time:.1f}s | inf={inference_ms:.2f}ms | mem={mem_mb:.1f}MB")

    # Kembalikan dictionary TANPA model_path (sesuai IndoBERT)
    return {
        "fold"            : fold_idx + 1,
        "qwk"             : final_qwk,
        "mse"             : final_mse,
        "training_time_s" : training_time,
        "inference_ms"    : inference_ms,
        "memory_mb"       : mem_mb,
        "memory_device"   : mem_device,
        "best_val_loss"   : best_val_loss,
        "n_epochs_actual" : n_epochs_actual,
        "val_idx"         : val_idx.tolist(),
        "train_idx"       : train_idx.tolist(),
        "history": {
            "train_loss": train_loss_history,
            "val_loss"  : val_loss_history,
            "val_qwk"   : val_qwk_history,
            "val_mse"   : val_mse_history,
        },
    }, fold_preds_df


# ══════════════════════════════════════════════════════════
#  MAIN
# ══════════════════════════════════════════════════════════
def main():
    set_seed(CFG["seed"])

    print("\n[1/5] Membaca dataset...")
    texts  = joblib.load(CFG["essay"])
    scores = joblib.load(CFG["score"])

    if os.path.exists(CFG["question_ids"]):
        question_ids = joblib.load(CFG["question_ids"])
    else:
        n_per_q = len(texts) // 10
        question_ids = np.repeat(np.arange(1, 11), n_per_q).astype(int)
        if len(question_ids) < len(texts):
            question_ids = np.append(question_ids,
                np.full(len(texts) - len(question_ids), 10, dtype=int))
        joblib.dump(question_ids, CFG["question_ids"])
        print(f"  question_ids dibuat dan disimpan (asumsi {n_per_q} esai/soal)")

    CFG["score_min"] = int(scores.min())
    CFG["score_max"] = int(scores.max())
    print(f"  Sampel: {len(texts)} | Skor: {CFG['score_min']}–{CFG['score_max']} | "
          f"Mean: {scores.mean():.2f}±{scores.std():.2f}")

    print("\n[2/5] Mengekstrak fitur linguistik...")
    fe_list = []
    for i, text in enumerate(texts):
        if (i+1) % 100 == 0 or i == 0:
            print(f"  [{i+1}/{len(texts)}]")
        fe_list.append(extract_linguistic_features(text))
    features_raw = np.array([[d[k] for k in FEATURE_NAMES] for d in fe_list],
                              dtype=np.float32)
    print(f"  Shape: {features_raw.shape} | Fitur: {FEATURE_NAMES}")

    print(f"\n[3/5] Memuat tokenizer: {CFG['model_name']}...")
    tokenizer = AutoTokenizer.from_pretrained(CFG["model_name"])

    print(f"\n[4/5] Memulai {CFG['n_folds']}-fold CV...")
    kf = KFold(n_splits=CFG["n_folds"], shuffle=True, random_state=CFG["seed"])
    all_results, all_preds_dfs = [], []

    for fold_idx, (train_idx, val_idx) in enumerate(kf.split(texts)):

        train_idx, es_idx = inner_split(train_idx, scores, fold_idx)  # [FIX-LEAKAGE]
        try:
            result, preds_df = run_fold(
                fold_idx, train_idx, val_idx,
                texts, features_raw, scores, question_ids,
                tokenizer, CFG, DEVICE, es_idx=es_idx)
            all_results.append(result)
            all_preds_dfs.append(preds_df)
        except Exception as e:
            print(f"  [ERROR] Fold {fold_idx+1}: {e}")
            traceback.print_exc()

    print(f"\n[5/5] Menyimpan hasil...")
    all_preds_df = pd.concat(all_preds_dfs, ignore_index=True)

    overall_qwk = quadratic_weighted_kappa(
        all_preds_df["true_score"].values,
        all_preds_df["pred_score"].values,
        CFG["score_min"], CFG["score_max"])
    overall_mse = mean_squared_error(
        all_preds_df["true_score"].values,
        all_preds_df["pred_score"].values)

    score_dist = {}
    for q in range(1, 11):
        mask = question_ids == q
        if mask.sum() > 0:
            score_dist[int(q)] = {
                "mean"  : float(scores[mask].mean()),
                "std"   : float(scores[mask].std()),
                "n"     : int(mask.sum()),
                "counts": {int(k): int(v) for k, v in
                           zip(*np.unique(scores[mask].astype(int),
                                          return_counts=True))},
            }

    experiment_data = {
        "config"            : CFG,
        "fold_results"      : all_results,
        "fold_predictions"  : all_preds_df,
        "overall_qwk"       : overall_qwk,
        "overall_mse"       : overall_mse,
        "features_raw"      : features_raw,
        "feature_names"     : FEATURE_NAMES,
        "scores"            : scores,
        "question_ids"      : question_ids,
        "score_distribution": score_dist,
        "model_name"        : CFG["model_name"],
        "paradigm"          : "fine-tuning",
        "n_samples"         : len(texts),
        "timestamp"         : datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    }

    joblib_path = os.path.join(CFG["output_dir"], "experiment_results.joblib")
    joblib.dump(experiment_data, joblib_path)

    results_df     = pd.DataFrame(all_results)
    results_df_csv = results_df.drop(columns=["history","val_idx","train_idx"],
                                      errors="ignore")
    results_df_csv.to_csv(
        os.path.join(CFG["output_dir"], "cv_results_distilbert_ft.csv"), index=False)
    all_preds_df.to_csv(
        os.path.join(CFG["output_dir"], "all_predictions_distilbert_ft.csv"), index=False)

    print(f"\n{'='*55}")
    print("  RINGKASAN — DistilBERT Fine-tuning + FE (sama dengan IndoBERT)")
    print(f"{'='*55}")
    print(results_df_csv.to_string(index=False))
    print(f"\n  Rata-rata ± Std:")
    for m in ["qwk","mse","training_time_s","inference_ms","memory_mb"]:
        print(f"    {m:<20s}: {results_df[m].mean():.4f} ± {results_df[m].std():.4f}")
    print(f"\n  QWK keseluruhan : {overall_qwk:.4f}")
    print(f"  MSE keseluruhan : {overall_mse:.4f}")
    print(f"\n  Disimpan di     : {joblib_path}")
    print(f"  Keys            : {list(experiment_data.keys())}")

    return experiment_data


if __name__ == "__main__":
    main()
```
